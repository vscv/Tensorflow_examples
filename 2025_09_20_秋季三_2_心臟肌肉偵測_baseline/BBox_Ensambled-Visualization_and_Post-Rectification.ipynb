```python
import pandas as pd
import matplotlib.pyplot as plt
import re
import os

# --- 1. 數據讀取與解析函式 ---

def load_yolo_submission(file_path):
    """
    讀取 YOLO 競賽提交檔，並解析成 DataFrame。
    欄位順序：[檔名, 類別, 信心指數, x1, y1, x2, y2]
    """
    try:
        # 由於數據沒有列名，我們手動指定它們
        df = pd.read_csv(
            file_path, 
            sep=' ', 
            header=None, 
            names=['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']
        )
        return df
    except FileNotFoundError:
        print(f"錯誤：找不到檔案 {file_path}")
        return None
    except Exception as e:
        print(f"讀取檔案時發生錯誤: {e}")
        return None

# --- 2. 數據處理與轉換函式 ---

def preprocess_data(df):
    """
    從 DataFrame 中提取病人ID和影像Z軸編號。
    """
    if df is None:
        return None

    # 提取病人ID (patient00##) 和 Z軸編號 (####)
    # 使用正規表達式提取：
    # patient(00\d{2}) -> 擷取 00xx 部分
    # _(\d{4}) -> 擷取四位數字部分
    df['patient_id'] = df['filename'].str.extract(r'(patient\d{4})')
    df['z_index'] = df['filename'].str.extract(r'_(\d{4})').astype(int)

    # 計算預測框的 Y 軸高度（用於視覺化）
    # y_center: 預測框的中心點 Y 座標
    # y_min: 預測框的最小 Y 座標 (y1)
    # y_max: 預測框的最大 Y 座標 (y2)
    df['y_min'] = df[['y1', 'y2']].min(axis=1) # 確保 y_min 總是較小的值
    df['y_max'] = df[['y1', 'y2']].max(axis=1) # 確保 y_max 總是較大的值
    df['y_center'] = (df['y_min'] + df['y_max']) / 2
    # X_
    df['x_min'] = df[['x1', 'x2']].min(axis=1)
    df['x_max'] = df[['x1', 'x2']].max(axis=1)
    df['x_center'] = (df['x_min'] + df['x_max']) / 2
    # W, H
    df['W'] = (df['x_max'] - df['x_min'])
    df['H'] = (df['y_max'] - df['y_min'])
    # score
    df['score'] = df['confidence'] * 10
    
    return df

# --- 3. 繪圖函式：Z軸高度分佈 ---

def plot_z_axis_distribution_ZY(df_processed, patient_id):
    """
    繪製特定病人的 Z軸 (影像編號) vs Y軸 (預測框上下線) 分佈圖。
    
    Args:
        df_processed (pd.DataFrame): 經過 preprocess_data 處理後的 DataFrame。
        patient_id (str): 要繪圖的病人ID (例如 'patient0091')。
    """
    if df_processed is None:
        return
        
    # 過濾出特定病人的數據
    patient_df = df_processed[df_processed['patient_id'] == patient_id].copy()

    if patient_df.empty:
        print(f"找不到病人ID: {patient_id} 的數據。")
        return

    # 確保按照 Z軸編號排序，以便繪圖線條連續
    patient_df = patient_df.sort_values(by='z_index')

    # 建立圖表
    fig, ax = plt.subplots(figsize=(15, 8))

    # 繪製每個預測框的 Y軸上下線作為一條垂直線
    # X軸: z_index (影像編號)
    # Y軸: y_min 到 y_max
    for index, row in patient_df.iterrows():
        # 繪製垂直線條 (Y軸的上下界線)
        # alpha=row['confidence'] 可以讓信心分數低的框更透明
        ax.plot(
            [row['z_index'], row['z_index']],  # Z軸座標
            [row['y_min'], row['y_max']],     # Y軸座標範圍
            color='blue', 
            alpha=row['confidence'],          # 根據信心分數調整透明度
            marker='_',                       # 在上下端點加一小橫線
            markersize=5,
            linewidth=1.5
        )
        
        # 繪製 Y 軸中心點
        ax.scatter(row['z_index'], row['y_center'], 
                   color='red', 
                   s=40 * row['confidence'] + 1, # 信心分數越高，點越大
                   zorder=5) # 確保中心點在最上層

    # 設定圖表標籤和標題
    ax.set_xlabel("Z-Axis (CT Image Index, e.g., 0150, 0151...)")
    ax.set_ylabel("Y-Axis Pixel Location (Top to Bottom)")
    ax.set_title(f"YOLO Bounding Box Y-Axis Distribution for Patient: {patient_id}")
    
    # 顯示網格線
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # 將 Y 軸反轉，通常在影像座標系中 Y=0 在頂部
    ax.invert_yaxis()
    
    plt.show()

    
    
    
def plot_z_axis_distribution_ZX(df_processed, patient_id):
    """
    繪製特定病人的 Z軸 (影像編號) vs X軸 (預測框左右邊線) 分佈圖。
    
    Args:
        df_processed (pd.DataFrame): 經過 preprocess_data 處理後的 DataFrame。
        patient_id (str): 要繪圖的病人ID (例如 'patient0091')。
    """
    if df_processed is None:
        return
        
    # 過濾出特定病人的數據
    patient_df = df_processed[df_processed['patient_id'] == patient_id].copy()

    if patient_df.empty:
        print(f"找不到病人ID: {patient_id} 的數據。")
        return

    # 確保按照 Z軸編號排序，以便繪圖線條連續
    patient_df = patient_df.sort_values(by='z_index')

    # 建立圖表
    fig, ax = plt.subplots(figsize=(15, 8))

    # 繪製每個預測框的 Y軸上下線作為一條垂直線
    # X軸: z_index (影像編號)
    # Y軸: y_min 到 y_max
    for index, row in patient_df.iterrows():
        # 繪製垂直線條 (Y軸的上下界線)
        # alpha=row['confidence'] 可以讓信心分數低的框更透明
        ax.plot(
            [row['z_index'], row['z_index']],  # Z軸座標
            [row['x_min'], row['x_max']],     # 切換成X軸座標範圍
            color='green', 
            alpha=row['confidence'],          # 根據信心分數調整透明度
            marker='_',                       # 在上下端點加一小橫線
            markersize=5,
            linewidth=1.5
        )
        
        # 繪製 Y 軸中心點
        ax.scatter(row['z_index'], row['x_center'], 
                   color='red', 
                   s=40 * row['confidence'] + 1, # 信心分數越高，點越大
                   zorder=5) # 確保中心點在最上層

    # 設定圖表標籤和標題
    ax.set_xlabel("Z-Axis (CT Image Index, e.g., 0150, 0151...)")
    ax.set_ylabel("X-Axis Pixel Location (Top to Bottom)")
    ax.set_title(f"YOLO Bounding Box X-Axis Distribution for Patient: {patient_id}")
    
    # 顯示網格線
    ax.grid(True, linestyle='--', alpha=0.6)
    
    # 將 Y 軸反轉，通常在影像座標系中 Y=0 在頂部
    ax.invert_yaxis()
    
    plt.show()
    
    
    
    
    
# --- 4. 執行流程範例 ---

# 1. 將您的數據樣本存為 'submission_sample.txt'
submission_file = DWBF_txt

# # 為了讓程式碼可立即運行，我們先將您的樣本數據寫入檔案
# sample_data = """patient0091_0200 0 0.6889 221 253 250 286
# patient0079_0200 0 0.6125 228 247 256 269
# patient0066_0180 0 0.7980 164 193 233 258
# patient0099_0242 0 0.8067 154 248 217 314
# patient0100_0244 0 0.7568 216 247 268 310
# patient0100_0244 0 0.0678 243 249 267 275
# patient0055_0217 0 0.7396 198 228 234 280
# patient0088_0215 0 0.3395 217 266 233 281
# patient0051_0263 0 0.7929 206 244 265 312
# patient0052_0156 0 0.7687 237 231 279 280
# patient0084_0162 0 0.8068 161 197 225 267
# patient0073_0194 0 0.6492 227 231 258 265
# patient0094_0206 0 0.5075 241 224 258 240
# patient0054_0234 0 0.7638 221 263 260 317
# patient0087_0164 0 0.7452 175 210 207 258
# patient0089_0188 0 0.6553 195 266 230 321
# patient0080_0185 0 0.8049 180 300 238 361
# patient0066_0174 0 0.7814 170 195 227 256
# patient0087_0184 0 0.7838 162 189 235 263
# patient0052_0178 0 0.7721 256 222 307 290
# patient0091_0196 0 0.6312 225 256 247 282
# patient0098_0201 0 0.8010 178 240 228 295
# patient0053_0174 0 0.5047 214 252 232 270
# patient0054_0242 0 0.7682 218 260 265 318
# patient0098_0199 0 0.7918 180 240 227 295
# patient0088_0194 0 0.7887 199 270 253 335
# """
# with open(submission_file, 'w') as f:
#     f.write(sample_data)


# 1. 載入數據
df_raw = load_yolo_submission(submission_file)

# 2. 處理數據
df_clean = preprocess_data(df_raw)

# # 3. 選擇病人 ID 進行繪圖
# # 您可以將這裡的 ID 換成您想檢查的任何病人 ID
# target_patient = 'patient0100'

# if df_clean is not None:
#     print(f"\n正在繪製 {target_patient} 的 Z軸高度分佈圖...")
#     plot_z_axis_distribution(df_clean, target_patient)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import math
from tqdm import tqdm


"""
Y軸座標改為： top-left

修正說明：計算全域 Y 軸範圍：在迴圈開始前，我們計算了所有數據中 $Y_{min}$ 和 $Y_{max}$ 的極值，並加入了 5% 的邊界。
移除 ax.invert_yaxis()：我們不再使用 ax.invert_yaxis()，因為它在共享軸上有時行為不穩定。使用 ax.set_ylim(max, min) 強制反轉：
matplotlib 允許您透過 ax.set_ylim(bottom, top) 設定 Y 軸的極限。如果您將最大值（例如 y_lim_max）作為底部 (bottom) 傳入，將最小值（例如 y_lim_min）作為頂部 (top) 傳入，
座標軸就會被強制反轉：Pythonax.set_ylim(y_lim_max, y_lim_min) 
這樣修改後，無論數據多寡或軸是否共享，所有子圖的 $Y$ 軸都將被統一設定並保持反轉，確保 $Y=0$ 在圖表的頂部。
"""
def batch_plot_z_axis_distribution_TL(df_processed, output_filename="all_patients_z_axis_distribution.png"):
    """
    自動繪製所有病人的 Z軸高度分佈圖，並將其合併到一張大圖中。
    """
    
    count_box = 0
    
    if df_processed is None:
        print("錯誤：輸入數據為 None，無法繪圖。")
        return

    # 1. 取得所有不重複的病人 ID，並按數字排序
    all_patients = sorted(df_processed['patient_id'].unique())
    num_patients = len(all_patients)

    if num_patients == 0:
        print("數據中沒有找到任何病人 ID。")
        return

    print(f"找到 {num_patients} 個病人 ID，正在準備繪製網格圖...")
    
    # --- 【新增步驟】計算所有病人的 Y 軸全域範圍 ---
    # 這是為了確保 Y 軸反轉和範圍設定在共享軸上能正確運作
    y_global_min = df_processed['y_min'].min()
    y_global_max = df_processed['y_max'].max()
    # 給予一些邊界空間
    y_range = y_global_max - y_global_min
#     y_lim_min = y_global_min - y_range * 0.05
#     y_lim_max = y_global_max + y_range * 0.05
    y_lim_min = 150   #統一設定Y軸上下限
    y_lim_max = 450   #統一設定Y軸上下限

    print(f"y_lim_min : {y_lim_min}, y_lim_max : {y_lim_max}")  # y_lim_min : 165.9, y_lim_max : 432.1
    
    # 2. 計算網格佈局大小
    cols = min(10, num_patients)
    rows = math.ceil(num_patients / cols)

    # 3. 建立大圖和所有子圖
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 2.5, rows * 2.5), 
                             sharey=True, # 保持共享 Y 軸
                             tight_layout=True)
    
    axes = axes.flatten() 

    # 4. 迭代每個病人並繪製到對應的子圖上
    for i, patient_id in tqdm(enumerate(all_patients), total=num_patients, desc="繪製病人圖表"):
        ax = axes[i]
        
        # 過濾出特定病人的數據
        patient_df = df_processed[df_processed['patient_id'] == patient_id].copy()
        patient_df = patient_df.sort_values(by='z_index')

        # 繪圖邏輯
        for _, row in patient_df.iterrows():
            # 繪製垂直線條 (Y軸的上下界線)
            ax.plot(
                [row['z_index'], row['z_index']],
                [row['y_min'], row['y_max']],
                color='blue', 
                alpha=max(0.2, row['confidence']),
                marker='_',
                markersize=3,
                linewidth=1
            )
            
            # 繪製 Y 軸中心點
            ax.scatter(row['z_index'], row['y_center'], 
                       color='red', 
                       s=4 * row['confidence'] + 0.01, 
                       zorder=5)
            
            count_box += 1

        # 設定子圖標題和標籤
        ax.set_title(patient_id, fontsize=10)
        ax.set_xlabel("Z-Index", fontsize=8)
        
        # 僅在最左側一列顯示 Y 軸標籤
        if i % cols == 0:
            ax.set_ylabel("Y-Axis", fontsize=8)
        
        # 調整刻度標籤大小
        ax.tick_params(axis='both', which='major', labelsize=7)
        ax.grid(True, linestyle=':', alpha=0.5)

        # --- 【修正處】強制設定 Y 軸範圍並反轉 ---
        # 即使共享 Y 軸，也強制設定範圍並反轉，確保所有子圖統一
        ax.set_ylim(y_lim_max, y_lim_min) # 注意：這裡的 max, min 順序是反的！
                                         # (y_lim_max 在第一個參數，讓它變成 Y 軸頂部)


    # 5. 隱藏多餘的子圖
    for j in range(num_patients, rows * cols):
        fig.delaxes(axes[j])

    # 6. 儲存影像
    print(f"正在儲存檔案至: {output_filename} \t總計: {count_box} box")
    plt.suptitle(f"{fold_name} \nZ-Axis VS Y-Axis Height Distribution Comparison Across All Patients", fontsize=16, y=1.02)
    plt.savefig(output_filename, dpi=300, bbox_inches='tight')
    plt.close(fig)
    plt.clf() #每次都清除內存避免累積記憶體用量
    print("所有病人的圖表已合併並儲存完畢。")
```

```python
%%time

# 假設 df_clean 已經在之前的步驟中成功生成。
save_all_patient_height_map = f"{DWBF_txt}_ZY_all_patient_height_map.png"
if df_clean is not None:
    batch_plot_z_axis_distribution_TL(df_clean, output_filename=save_all_patient_height_map)
```

# Z軸投影


```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import math
from tqdm import tqdm
import matplotlib.patches as patches
import os

# 假設 load_yolo_submission 和 preprocess_data 已經在您的環境中定義。

def batch_plot_xy_projection_map(df_processed, output_filename="all_patients_xy_projection_map.png", fold_name="", image_size=512):
    """
    【批量 XY 累積投影圖】
    繪製每個病人單獨的 (X, Y) 累積投影圖，並將其合併到一張大圖中。
    每個框以矩形表示，並以紅點標示其形心。
    
    Args:
        df_processed (pd.DataFrame): 已經過 preprocess_data 處理後的 DataFrame。
        output_filename (str): 輸出圖表的檔案名稱。
        fold_name (str): 用於圖表總標題的資料夾/批次名稱。
        image_size (int): 影像的單邊像素大小 (通常為 512)。
    """
    
    if df_processed is None:
        print("錯誤：輸入數據為 None，無法繪圖。")
        return

    # 1. 取得所有不重複的病人 ID，並按數字排序
    all_patients = sorted(df_processed['patient_id'].unique())
    num_patients = len(all_patients)

    if num_patients == 0:
        print("數據中沒有找到任何病人 ID。")
        return

    print(f"找到 {num_patients} 個病人 ID，正在準備繪製 (X, Y) 投影網格圖...")
    
    # 2. 計算網格佈局大小
    cols = min(10, num_patients)
    rows = math.ceil(num_patients / cols)

    # 3. 建立大圖和所有子圖
    # 我們不需要 sharey/sharex，因為每個子圖的 X/Y 軸都應該是獨立的 [0, 512] 範圍。
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 2.5, rows * 2.5), 
                             tight_layout=True)
    
    # 處理單行或單列的情況
    axes = axes.flatten() 

    total_boxes = 0

    # 4. 迭代每個病人並繪製到對應的子圖上
    for i, patient_id in tqdm(enumerate(all_patients), total=num_patients, desc="繪製病人 XY 投影圖"):
        ax = axes[i]
        
        # 過濾出特定病人的數據
        patient_df = df_processed[df_processed['patient_id'] == patient_id].copy()
        num_patient_boxes = len(patient_df)
        total_boxes += num_patient_boxes

        # 繪圖邏輯：迭代該病人的所有偵測框
        for _, row in patient_df.iterrows():
            
            # 繪製偵測框矩形 (Bounding Box)
            x_min = row['x_min']
            y_min = row['y_min']
            width = row['x_max'] - row['x_min']
            height = row['y_max'] - row['y_min']
            
            rect = patches.Rectangle(
                (x_min, y_min), # 左上角座標
                width, 
                height,
                linewidth=0.5,
                edgecolor='blue', 
                facecolor='none', 
                alpha=max(0.1, row['confidence'] * 0.5), # 透明度
                zorder=1
            )
            ax.add_patch(rect)
            
            # 繪製形心 (Center Point)
            ax.scatter(
                row['x_center'], 
                row['y_center'], 
                color='red', 
                marker='.',
                s=5 * row['confidence'] + 0.5, # 信心分數影響大小
                alpha=max(0.2, row['confidence']),
                zorder=2
            )

        # 5. 設定子圖參數
        
        ax.set_title(f"{patient_id} ({num_patient_boxes} boxes)", fontsize=8)
        
#         # 設定 X, Y 軸範圍
#         ax.set_xlim(150, 450)
#         ax.set_ylim(150, 450)
        
        
        
        # Y 軸反轉：Y=0 在頂部
        ax.set_ylim(image_size, 0) 
        
        ax.set_aspect('equal', adjustable='box') # 確保 X, Y 軸比例為 1:1
        
        # 僅在最底排和最左側一列顯示軸標籤
        if i >= (rows - 1) * cols:
            ax.set_xlabel("X-Axis", fontsize=7)
        else:
            ax.tick_params(labelbottom=False) # 隱藏 X 軸標籤

        if i % cols == 0:
            ax.set_ylabel("Y-Axis", fontsize=7)
        else:
            ax.tick_params(labelleft=False) # 隱藏 Y 軸標籤

        ax.tick_params(axis='both', which='major', labelsize=6)
        ax.grid(True, linestyle=':', alpha=0.3)
        
        
        # --- 【修正處】強制設定 Y 軸範圍並反轉 ---
        # 即使共享 Y 軸，也強制設定範圍並反轉，確保所有子圖統一
        #ax.set_ylim(x_lim_max, x_lim_min) # 注意：這裡的 max, min 順序是反的！
        ax.set_xlim(150, 300)
        ax.set_ylim(350, 150)
        
        
    # 6. 隱藏多餘的子圖
    for j in range(num_patients, rows * cols):
        fig.delaxes(axes[j])

    # 7. 儲存影像
    print(f"總計繪製 {total_boxes} 個框。")
    plt.suptitle(f"{fold_name} \nAccumulated (X, Y) Bounding Box Projection Map by Patient", fontsize=16, y=1.02)
    plt.savefig(output_filename, dpi=300, bbox_inches='tight')
    plt.close(fig)
    plt.clf()
    print("所有病人的 (X, Y) 投影網格圖已儲存完畢。")

# --- 範例執行流程 ---
# # 假設您已經定義了 load_yolo_submission 和 preprocess_data
# cleaned_file = "DETR_1120_V2_AR4_clean_v2.txt"

# # 1. 讀取清理後的數據並預處理 (獲取 x_min, x_max, x_center 等)
# df_raw_cleaned = load_yolo_submission(cleaned_file)
# df_cleaned_processed = preprocess_data(df_raw_cleaned) 

# # 2. 批量繪製 (X, Y) 累積投影圖
# if df_cleaned_processed is not None:
#     batch_plot_xy_projection_map(
#         df_cleaned_processed, 
#         output_filename="DETR_1120_V2_AR4_clean_XY_projection_batch.png",
#         fold_name="DETR_1120_V2_AR4_clean_V2"
#     )
```

# 聚類分析


工作原理
定義鄰域：給定一個鄰域半徑 eps 和最小點數 min_samples，算法會檢查每個數據點周圍有多少個點在 eps 距離範圍內。
核心點、邊界點和噪音點：2-1 核心點：如果一個數據點周圍有超過 min_samples 個點在 eps 範圍內，則該點是核心點。2-2 邊界點：不滿足核心點條件，但在其他核心點的 eps 鄰域內的點。2-3 噪音點：既不是核心點，也不在任何核心點的鄰域內的點，標記為噪音點（label = -1）。
形成群集：從一個核心點開始，將所有相連的點（包括邊界點）歸入同一群集，直到不再有新的點可以加入該群集為止。

# DB clean V3
##### 接續步驟，另外寫一個新的處理function同樣使用已經處理好的df_clean當作輸入，針對patient0061的分群結果，不論分析後有多少群組(通常為2~4群)，若有兩群在Z軸位置上重疊超過20個框時，對其重疊的區域中Y軸值較大的群組，從x軸向前刪除12個框。繪製該病人子圖即可不用大圖，列印被刪除多少重複點，最後另存新的DBSCAN_clean_1129_v3.txt檔案。

```python
def process_patient0061_clusters(df_clean, DBSCAN_updated_V3_name='DBSCAN_clean_1129_v3.txt'):
    """
    新處理函式：針對 patient0061 的聚類結果進行刪除
    """
    from sklearn.cluster import DBSCAN
    import matplotlib.pyplot as plt
    
    # 提取 patient0061
    patient_data = df_clean[df_clean['patient_id'] == 'patient0061'].copy()
    patient_data = patient_data.sort_values('z_index')
    
    if patient_data.empty:
        print("patient0061 無數據")
        return df_clean
    
    original_count = len(patient_data)
    print(f"原始點數：{original_count}")
    
    # 聚類 (X=z_index, Y=y_center)
    X = patient_data['z_index'].values
    Y = patient_data['y_center'].values
    points = np.column_stack((X, Y))
    
    db = DBSCAN(eps=4, min_samples=1).fit(points)
    labels = db.labels_
    
    # 群組統計 (忽略噪聲 -1)
    clusters = {}
    for i, label in enumerate(labels):
        if label != -1:
            if label not in clusters:
                clusters[label] = []
            clusters[label].append(i)
    
    print(f"群組數：{len(clusters)}")
    for label, idxs in clusters.items():
        print(f"  簇 {label}: {len(idxs)} 點")
    
    # 檢查重疊 (任意兩群 Z 重疊 >20)
    overlap_groups = []
    for label1 in clusters:
        for label2 in clusters:
            if label1 >= label2:
                continue
            z1 = set(X[clusters[label1]])
            z2 = set(X[clusters[label2]])
            overlap_z = len(z1.intersection(z2))
            if overlap_z > 10:
                overlap_groups.append((label1, label2, overlap_z))
    
    if not overlap_groups:
        print("無重疊群組 >20，無刪除")
        return df_clean
    
    print(f"重疊群組：{overlap_groups}")
    
    # 選 Y 平均較大的群組，從其最小 Z 刪除  後 12 個
    deleted_count = 0
    for label1, label2, _ in overlap_groups:
        # 計算平均 Y
        y_avg1 = np.mean(Y[clusters[label1]])
        y_avg2 = np.mean(Y[clusters[label2]])
        
        # 選 Y 較大的群組
        larger_label = label1 if y_avg1 > y_avg2 else label2
        larger_idxs = clusters[larger_label]
        
        # 從最小 Z 開始刪  後 12 個
        sorted_larger = sorted(larger_idxs, key=lambda i: X[i])
        # del 50% of overlap
        cut_n = int(len(sorted_larger) / 2)
        to_delete = sorted_larger[-cut_n:]
        
        # 從 patient_data 刪除這些索引
        patient_data = patient_data.drop(patient_data.index[to_delete])
        deleted_count += len(to_delete)
    
    print(f"被刪除重複點：{deleted_count} {cut_n}")
    
    # 更新 df_clean (只改 patient0061)
    df_clean.loc[df_clean['patient_id'] == 'patient0061', :] = np.nan  # 先清空
    updated_df = pd.concat([df_clean, patient_data], ignore_index=True)
    
    # 繪製子圖
    fig, ax = plt.subplots(1, 1, figsize=(10, 6))
    # 原始
    ax.scatter(X, Y, c='blue', s=30, alpha=0.6, label='Original')
    # 聚類
    colors = ['red' if l == -1 else f'C{l}' for l in labels]
    ax.scatter(X, Y, c=colors, s=40, edgecolors='black', linewidth=0.5, label='Clustered')
    # 刪除後
    remaining_X = patient_data['z_index'].values
    remaining_Y = patient_data['y_center'].values
    ax.scatter(remaining_X, remaining_Y, c='green', s=50, marker='^', label='After Deletion')
    ax.set_title('patient0061: Z vs Y_center (Original + Clustered + Deleted)')
    ax.set_xlabel('Z_index')
    ax.set_ylabel('Y_center')
    ax.set_ylim(350, 180)
    ax.grid(True, alpha=0.3)
    ax.legend()
    plt.tight_layout()
    plt.savefig('patient0061_updated_plot.png', dpi=150)
    plt.show()
    
#     # 儲存新 txt
#     updated_df = df_clean.sort_values('confidence', ascending=False)
#     for col in ['x1', 'y1', 'x2', 'y2']:
#         updated_df[col] = updated_df[col].round().astype(int)
#     updated_df[['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']].to_csv(output_txt, sep=' ', index=False, header=False, float_format='%.4f')
    
#     print(f"新檔案儲存：{output_txt} (總行數 {len(updated_df)})")

    # 儲存新 txt（修正版：刪除 NaN 行，避免 round/int 錯誤）
    updated_df = updated_df.sort_values('confidence', ascending=False)

    # Step 1: 刪除有 NaN 的行（直接移除，不填充 0）
    updated_df = updated_df.dropna(subset=['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2'])

    # Step 2: 只對剩餘的座標欄位進行 round 和轉型（確保無 NaN）
    for col in ['class', 'x1', 'y1', 'x2', 'y2']:
        updated_df[col] = updated_df[col].round().astype(int)

    # Step 3: 選取欄位並儲存
    output_cols = ['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']
    updated_df[output_cols].to_csv(DBSCAN_updated_V3_name, sep=' ', index=False, header=False, float_format='%.4f')

    print(f"儲存完成！總行數：{len(updated_df)}（已刪除 {original_count - len(updated_df)} 個 NaN 行）")
    print(f"檔案：{DBSCAN_updated_V3_name}")
    return df_clean


# 使用方式
DBSCAN_updated_V3_name = 'DBSCAN_clean_1130_v3.txt'
df_clean = process_patient0061_clusters(df_clean, DBSCAN_updated_V3_name)
```

這是一個極具實務價值且複雜的挑戰，涉及到**時序一致性（Temporal Consistency）和空間連續性（Spatial Continuity）**的處理，這在醫學影像（尤其是 CT/MRI 序列）的物件偵測中是從 $0.966$ 邁向 $0.975$ 甚至更高的關鍵步驟。由於 CT 序列影像在物理上是連續的（例如，心臟瓣膜在相鄰切片上的位置變化很小且平滑），我們需要對您通過 WBF 得到的邊界框結果進行序列化後處理。以下是針對 CT 序列影像偵測結果進行**整流（Rectification）、補缺（Infilling）、擴充（Extension）**以確保物理連續性的方法與步驟。序列影像邊界框後處理方法：時序平滑化我們將使用**跟蹤（Tracking）和平滑（Smoothing）**技術來達成目標。步驟零：數據準備假設您目前的 WBF 輸出結果 (wbf_ensemble_results_two_stage.txt) 已經是每個影像最多只有一個瓣膜預測，且格式為：$$\text{圖片名稱} \quad \text{類別} \quad \text{置信度} \quad x_{\min} \quad y_{\min} \quad x_{\max} \quad y_{\max}$$你需要將這些結果按照CT 序列的物理順序進行排序。通常圖片名稱中會包含切片序號（例如 patient0051_0222 到 patient0051_0231）。步驟一：跟蹤與數據結構化（Tracking & Structuring）在您的單目標情境中，跟蹤的目標是建立一個序列化的時間線。分組： 根據病人 ID（patient0051）將影像分組。時序排序： 在每個組內，根據切片號（0222, 0223, ...）對預測結果進行排序。定義狀態： 每個切片 $t$ 的狀態 $Z_t$ 是一個向量，包含：$\begin{pmatrix} C_t & S_t & B_{t,\min} & B_{t,\max} \end{pmatrix}$，其中 $C_t$ 是中心點座標 $(x_c, y_c)$，$S_t$ 是置信度，$B$ 是邊界框座標。步驟二：整流與補缺（平滑與內插）這是確保連續性的核心步驟，目的在於處理單個切片上因模型不確定性導致的跳動（Jumping）或遺漏（Missing）。1. 整流（邊界框平滑化）我們使用平滑濾波器來確保邊界框中心點和尺寸在序列上的平滑變化。技術： 卡爾曼濾波器（Kalman Filter） 或 指數移動平均（Exponential Moving Average, EMA）。實作： 對於每一張影像 $t$ 的中心點 $(x_{c,t}, y_{c,t})$ 和尺寸 $(w_t, h_t)$，我們計算一個平滑後的結果 $(\hat{x}_{c,t}, \hat{y}_{c,t}, \hat{w}_t, \hat{h}_t)$。EMA 示例：$$\hat{X}_t = \alpha \cdot X_t + (1 - \alpha) \cdot \hat{X}_{t-1}$$其中 $X$ 可以是 $x_c, y_c, w, h$ 或置信度，$\alpha$ 是平滑因子（$0 < \alpha < 1$）。$\alpha$ 越小，平滑效果越強，但延遲越大。2. 補缺（Infilling Missing Slices）當某些切片因為置信度低於閾值而被 WBF 篩除（預測結果為空）時，我們需要根據前後切片的預測來內插。方法： 線性內插（Linear Interpolation）。實作： 假設切片 $t$ 的預測遺失，但切片 $t-k$ 和 $t+m$ 有有效的平滑後預測 $\hat{X}_{t-k}$ 和 $\hat{X}_{t+m}$。我們對遺失的切片 $t, t+1, \dots, t+m-1$ 進行線性內插：$$\hat{X}_i = \hat{X}_{t-k} + (\hat{X}_{t+m} - \hat{X}_{t-k}) \cdot \frac{i - (t-k)}{m+k}$$置信度處理： 對於內插的框，其置信度應設定為一個較低的、但非零的值（例如 $\text{MIN\_CONFIDENCE}$），或直接採用內插值。步驟三：擴充（Extending Boundary）這個步驟旨在利用物理連續性來擴展有效的預測範圍。有效區段定義： 找出連續具有高置信度預測（例如置信度 $> 0.8$）的切片範圍 $[t_{start}, t_{end}]$。邊界擴展： 如果 $t_{start}-1$ 或 $t_{end}+1$ 恰好是沒有預測的切片，我們嘗試將有效區段的邊界擴展到這些切片上。方法： 使用邊界切片（$t_{start}$ 或 $t_{end}$）的平滑化預測 $\hat{X}$ 作為外插的基礎，並將置信度逐步遞減。目的： 透過平滑化後的數據，可以在模型不確定但在物理上應存在的邊緣切片上生成預測框。

```python
# --- [設定參數] ---
ALPHA = 0.5            
MIN_CONFIDENCE_INF = 0.05 

def temporal_smoothing_and_infilling(wbf_results_df):
    """
    對 WBF 輸出的邊界框序列進行時序平滑和補缺。
    
    Args:
        wbf_results_df (pd.DataFrame): 包含 WBF 結果的 DataFrame，
                                       必須包含 'filename', 'confidence', 'z_index' 等欄位。
    """
    
    # 1. 準備數據：選取需要平滑化的特徵
    
    # *** 關鍵修復：將 'confidence' 重命名為 'score'，並確保後續代碼使用 'score' ***
    df = wbf_results_df.rename(columns={'filename': 'image_name', 'confidence': 'score'})
    
    # 由於您的 x1, y1, x2, y2 已經是像素座標，我們計算寬度和高度
    df['w'] = df['x2'] - df['x1']
    df['h'] = df['y2'] - df['y1']
    
    # *** 關鍵修復：data_cols 列表必須使用 'score' 欄位名稱 ***
    data_cols = ['x_center', 'y_center', 'score', 'w', 'h'] 
    
    # 創建一個用於輸出的新 DataFrame
    smoothed_df = df[['image_name', 'patient_id', 'z_index', 'class']].copy()
    
    # ... (其餘代碼保持不變)
    
    # 2. 函數：執行 EMA 平滑
    def apply_ema(series, alpha):
        """ 使用指數移動平均進行平滑 """
        return series.ewm(alpha=alpha, adjust=False).mean()

    # 3. 按病人ID分組並執行平滑和內插
    
    # 由於我們在 enforce_single_box_per_image 中已經處理過，這裡只需要處理剩下的邏輯
    
    # 處理每一個病人序列
    for patient_id, group in df.groupby('patient_id'):
        
        # 3.1 確保序列完整性 (關鍵步驟)
        min_z = group['z_index'].min()
        max_z = group['z_index'].max()
        full_z_range = pd.Series(range(min_z, max_z + 1), name='z_index')
        
        full_sequence = full_z_range.to_frame().merge(
            group.set_index('z_index'), 
            left_on='z_index', 
            right_index=True, 
            how='left'
        ).sort_values('z_index')

        # 3.2 執行整流 (EMA 平滑)
        for col in data_cols:
            # 只有非 NaN 的數據才會參與 EMA 計算
            smoothed_values = apply_ema(full_sequence[col].dropna(), ALPHA)
            
            # 將平滑後的結果放回 full_sequence 的對應位置
            full_sequence.loc[smoothed_values.index, f'smoothed_{col}'] = smoothed_values
            
            # 使用線性內插來填充缺失值 (補缺)
            full_sequence[f'smoothed_{col}'] = full_sequence[f'smoothed_{col}'].interpolate(method='linear')

        # 3.3 最終輸出整理
        
        # 限制內插的置信度 (score)
        full_sequence['smoothed_score'] = full_sequence['smoothed_score'].clip(lower=MIN_CONFIDENCE_INF)
        
        # 處理沒有原始預測但被內插的切片：繼承其他資訊
        full_sequence['image_name'] = full_sequence['image_name'].fillna(method='ffill').fillna(method='bfill')
        full_sequence['patient_id'] = full_sequence['patient_id'].fillna(patient_id)
        full_sequence['class'] = full_sequence['class'].fillna(0).astype(int) 

        # 將處理後的數據附加到最終結果
        # 這裡需要小心處理索引，避免重複 append 造成資料膨脹
        smoothed_df = pd.concat([smoothed_df, full_sequence])


    # 4. 重新計算平滑後的邊界框
    
    # 由於我們在循環中使用了 pd.concat，這裡需要重新處理索引和重複行
    final_output = smoothed_df[smoothed_df['patient_id'].notna()].drop_duplicates(subset=['image_name', 'z_index'], keep='last')
    final_output = final_output.sort_values(['patient_id', 'z_index']).reset_index(drop=True)
    
    # 計算邊界框座標 (使用四捨五入轉換回像素整數)
    final_output['xmin_s'] = (final_output['smoothed_x_center'] - final_output['smoothed_w'] / 2).round().astype('Int64')
    final_output['xmax_s'] = (final_output['smoothed_x_center'] + final_output['smoothed_w'] / 2).round().astype('Int64')
    final_output['ymin_s'] = (final_output['smoothed_y_center'] - final_output['smoothed_h'] / 2).round().astype('Int64')
    final_output['ymax_s'] = (final_output['smoothed_y_center'] + final_output['smoothed_h'] / 2).round().astype('Int64')
    
#     # 5. 選擇最終輸出欄位
#     return final_output[[
#         'image_name', 'class', 'smoothed_score', 'xmin_s', 'ymin_s', 'xmax_s', 'ymax_s',
#         'patient_id', 'z_index'
#     ]].rename(columns={
#         'smoothed_score': 'score',
#         'xmin_s': 'x1_s',
#         'ymin_s': 'y1_s',
#         'xmax_s': 'x2_s',
#         'ymax_s': 'y2_s'
#     })

    # 5. 選擇最終輸出欄位 (只留原本的 檔名 分數 左上右下座標)
    return final_output[[
        'image_name', 'class', 'smoothed_score', 'xmin_s', 'ymin_s', 'xmax_s', 'ymax_s'
    ]].rename(columns={
        'smoothed_score': 'score',
        'xmin_s': 'x1_s',
        'ymin_s': 'y1_s',
        'xmax_s': 'x2_s',
        'ymax_s': 'y2_s'
    })
```

# 步驟：修復 WBF 輸出並優化時序程式

您好，感謝您的即時回饋與細心觀察！如果您的 WBF 集成結果文件 (wbf_ensemble_two_stage_results.txt) 中，某些影像仍然包含兩個或兩個以上的邊界框，那麼這是一個非常嚴重的問題，它直接違背了我們在兩階段 WBF 集成中建立的**「單目標約束」，並且會導致時序平滑化程式無法直接使用**。為什麼時序平滑化程式無法直接使用？我們之前設計的「序列影像邊界框後處理方法：時序平滑化」程式是基於一個強烈假設：假設： 每個切片 $t$ 最多只有一個預測框 $B_t$（即單目標）。如果一張圖有 $N>1$ 個預測框 $B_{t,1}, B_{t,2}, \dots, B_{t,N}$，我們就無法確定：這些框是否屬於同一個物體？ (瓣膜)應該對哪個框進行時序平滑？在時序平滑和補缺時，應考慮哪個框的歷史軌跡？因此，在進行時序平滑之前，我們必須先強制執行單目標約束。解決方案：在時序處理前強制單目標約束由於這個問題發生在兩階段 WBF 之後，這意味著：階段一的 WBF 淨化（單模型內部淨化並保留最高分）可能失敗了，或階段二的跨模型 WBF 融合（將 $K$ 個單框融合）在某些情況下輸出超過一個框（這意味著 K 個模型對兩個不同的位置或物體達成了共識）。無論原因為何，解決方案都是在時序平滑化之前，加入一個最終的、最嚴格的**「最高置信度單框篩選」**。步驟：修復 WBF 輸出並優化時序程式1. 最終強制單目標篩選（修復 WBF 輸出）我們需要在讀取 WBF 輸出結果時，對每一張圖強制只保留置信度最高的一個框。

總結加入` enforce_single_box_per_image `函數是必要的安全閥。它強制將您的 WBF 輸出轉換回 $N=1$ 或 $N=0$ 的單目標格式，從而滿足時序平滑演算法對單一軌跡輸入的嚴格要求。請將此修復步驟加入您的工作流程中，然後再繼續進行時序平滑化。

```python
# # 修復步驟：在時序處理前，對 WBF 輸出結果進行單目標強制篩選

# def enforce_single_box_per_image(wbf_results_df):
#     """
#     對 WBF 輸出的結果進行最終的單目標約束：每張圖只保留置信度最高的一個框。
#     """
    
#     # 按圖片名稱分組，並在每個組內按分數降序排序
#     # 然後只取每個組的第一行 (即最高分數的框)
    
#     # 確保 image_name 是唯一的分組鍵
#     wbf_results_df['image_name'] = wbf_results_df['image_name'].astype(str)
    
#     # 找出每張圖置信度最高的框
#     idx = wbf_results_df.groupby(['image_name'])['score'].idxmax()
    
#     # 篩選出這些最高的框
#     final_single_box_df = wbf_results_df.loc[idx].reset_index(drop=True)
    
#     return final_single_box_df
```

```python
# 修復步驟：在時序處理前，對 WBF 輸出結果進行單目標強制篩選

def enforce_single_box_per_image(wbf_results_df):
    """
    對 WBF 輸出的結果進行最終的單目標約束：每張圖只保留置信度最高的一個框。
    
    Args:
        wbf_results_df (pd.DataFrame): 包含 WBF 結果的 DataFrame，
                                       必須包含 'filename' (影像名稱) 和 'confidence' (置信度)。
        
    Returns:
        pd.DataFrame: 每張影像只包含置信度最高的一個邊界框。
    """
    
    # 確保 filename 是唯一的分組鍵
    # 這裡使用 'filename' 欄位作為影像名稱
    wbf_results_df['filename'] = wbf_results_df['filename'].astype(str)
    
    # 找出每張圖置信度最高的框
    # 這裡使用 'confidence' 欄位作為分數
    idx = wbf_results_df.groupby(['filename'])['confidence'].idxmax()
    
    # 篩選出這些最高的框
    final_single_box_df = wbf_results_df.loc[idx].reset_index(drop=True)
    
    # 為了保持後續程式碼的命名一致性，這裡可以考慮將 confidence 欄位更名為 score
    # 但為了與您的原始 DataFrame 保持一致，我們暫時保留 'confidence'。

    print(f"✅ 完成單目標篩選：影像數量從 {len(wbf_results_df.index)} 條紀錄減少到 {len(final_single_box_df.index)} 條單目標紀錄。")
    
    return final_single_box_df
```

```python
def natural_sort_key(s):
    """
    將字串分解為文字和數字列表，用於自然數字排序。
    s 必須是數據行 (e.g., 'patient0091_0200 0 0.6889 ...')
    """
    # 提取第一個欄位 (e.g., 'patient0091_0200')
    filename = s.split()[0]
    
    # 使用正規表達式(\d+)將字串分割成數字部分和非數字部分
    # 例如: 'patient0091_0200' -> ['patient', '0091', '_', '0200', '']
    return [int(text) if text.isdigit() else text.lower() 
            for text in re.split('(\d+)', filename)]


def sort_submission_file(input_path, output_path):
    # 讀取檔案
    try:
        with open(input_path, 'r') as f:
            lines = [line.strip() for line in f if line.strip()]
    except FileNotFoundError:
        print(f"錯誤：找不到檔案 {input_path}")
        return

    # 排序
    sorted_lines = sorted(lines, key=natural_sort_key)

    # 寫入新檔案
    with open(output_path, 'w') as f:
        for line in sorted_lines:
            f.write(line + '\n')
            
    print(f"檔案已成功排序並儲存到 {output_path}")

# 範例使用
# sort_submission_file('submission.txt', 'sorted_submission.txt')
```

```python
import os
import re

# --- 配置參數 ---
INPUT_DIR = "training_label"
OUTPUT_FILE = "training_labels_512.txt"
IMAGE_SIZE = 512  # 目標影像尺寸 (512x512)

# --- 座標轉換函式 ---

def yolo_to_abs_coords(x_center, y_center, width, height, img_size):
    """
    將 YOLO 歸一化座標轉換為絕對像素座標 (x_min, y_min, x_max, y_max)。
    並確保座標轉換為整數。
    讀取訓練標註檔，轉換 YOLO 座標為 512x512 絕對座標，並彙整輸出。
    同時統計每個子資料夾 (病人) 的 TXT 檔案數量。
    """
    # 轉換為絕對座標
    abs_x_center = x_center * img_size
    abs_y_center = y_center * img_size
    abs_width = width * img_size
    abs_height = height * img_size
    
    # 計算 (x_min, y_min, x_max, y_max)
    x_min = int(round(abs_x_center - abs_width / 2))
    y_min = int(round(abs_y_center - abs_height / 2))
    x_max = int(round(abs_x_center + abs_width / 2))
    y_max = int(round(abs_y_center + abs_height / 2))
    
    # 邊界保護：確保座標在 [0, img_size-1] 範圍內
    x_min = max(0, x_min)
    y_min = max(0, y_min)
    x_max = min(img_size - 1, x_max)
    y_max = min(img_size - 1, y_max)
    
    return x_min, y_min, x_max, y_max

# --- 主程序 ---

def aggregate_and_convert_labels():
    """
    讀取訓練標註檔，轉換 YOLO 座標為 512x512 絕對座標，並彙整輸出。
    """
    total_annotations = 0
    output_lines = []

    print(f"開始遍歷資料夾：{INPUT_DIR}...")
    
    # 遍歷所有子資料夾和檔案
    for root, _, files in os.walk(INPUT_DIR):
        for filename in files:
            if not filename.endswith('.txt'):
                continue
            
            file_path = os.path.join(root, filename)
            
            # 從檔名中提取 patientID 和 Z_index (e.g., patient0050_0242)
            # 這裡假設檔名格式是 'patientXXXX_YYYY.txt'
            base_name, _ = os.path.splitext(filename)
            
            # 讀取單個 TXT 檔案
            try:
                with open(file_path, 'r') as f:
                    for line in f:
                        line = line.strip()
                        if not line:
                            continue
                        
                        parts = line.split()
                        
                        # 檢查 YOLO 格式是否正確 (至少5個欄位: class x_c y_c w h)
                        if len(parts) < 5:
                            print(f"警告：檔案 {file_path} 中有格式錯誤的行：{line}")
                            continue
                        
                        try:
                            # 原始 YOLO 數據 (第0欄位是 class_id, 但在輸出時被固定為 0)
                            # 這裡只需要讀取座標，但我們保留 class_id 以備檢查
                            
                            # 注意：您的輸出格式中 class 欄位被固定為 0
                            # 因此我們使用原始的 class_id 作為參考，但輸出固定為 0 1
                            
                            # YOLO 座標是第 1 到 4 欄
                            x_c = float(parts[1])
                            y_c = float(parts[2])
                            w = float(parts[3])
                            h = float(parts[4])
                            
                            # 轉換座標
                            x_min, y_min, x_max, y_max = yolo_to_abs_coords(x_c, y_c, w, h, IMAGE_SIZE)
                            
                            # 格式化輸出行:
                            # 格式: patientID_Zindex 0 1 x_min y_min x_max y_max
                            # 這裡的 '0 1' 是根據您的需求固定的
                            output_line = f"{base_name} 0 1 {x_min} {y_min} {x_max} {y_max}"
                            
                            output_lines.append(output_line)
                            total_annotations += 1
                            
                        except ValueError:
                            print(f"警告：檔案 {file_path} 中有非數字格式錯誤的行：{line}")
                            continue

            except Exception as e:
                print(f"讀取檔案 {file_path} 時發生異常: {e}")
                continue

    # 將所有彙整的內容寫入單一檔案
    print(f"\n總共處理了 {total_annotations} 條標註。")
    try:
        with open(OUTPUT_FILE, 'w') as out_f:
            for line in output_lines:
                out_f.write(line + '\n')
        print(f"✅ 成功將結果寫入彙整檔案：{OUTPUT_FILE}")
    except Exception as e:
        print(f"寫入彙整檔案時發生錯誤: {e}")

# 執行主函式
aggregate_and_convert_labels()
```

# 統計數量＋排序輸出

```python
import os
import re
import math
import pandas as pd
from tqdm import tqdm # 為了美觀，可以保留 tqdm

# --- 配置參數 ---
INPUT_DIR = "training_label"
OUTPUT_FILE = "training_labels_512_reorder_V2.txt"
IMAGE_SIZE = 512  # 目標影像尺寸 (512x512)

# --- 座標轉換函式 (保持不變) ---
def yolo_to_abs_coords(x_center, y_center, width, height, img_size):
    """
    將 YOLO 歸一化座標轉換為絕對像素座標 (x_min, y_min, x_max, y_max)。
    """
    abs_x_center = x_center * img_size
    abs_y_center = y_center * img_size
    abs_width = width * img_size
    abs_height = height * img_size
    
    x_min = int(round(abs_x_center - abs_width / 2))
    y_min = int(round(abs_y_center - abs_height / 2))
    x_max = int(round(abs_x_center + abs_width / 2))
    y_max = int(round(abs_y_center + abs_height / 2))
    
    # 邊界保護
    x_min = max(0, x_min)
    y_min = max(0, y_min)
    x_max = min(img_size - 1, x_max)
    y_max = min(img_size - 1, y_max)
    
    return x_min, y_min, x_max, y_max

# --- 【新增】自然排序 Key 函式 ---
def natural_sort_key(s):
    """
    將字串分解為文字和數字列表，用於自然數字排序。
    排序依據是行首的檔名 (patient####_####)。
    """
    # 提取第一個欄位 (e.g., 'patient0091_0200')
    filename = s.split()[0]
    
    # 使用正規表達式(\d+)將字串分割成數字部分和非數字部分
    return [int(text) if text.isdigit() else text.lower() 
            for text in re.split('(\d+)', filename)]

# --- 主程序 (新增統計與排序功能) ---

def aggregate_and_convert_labels():
    """
    讀取訓練標註檔，轉換 YOLO 座標為 512x512 絕對座標，
    彙整輸出並進行自然排序，同時統計每個病人的 TXT 檔案數量。
    """
    total_annotations = 0
    output_lines = []
    folder_txt_counts = {}

    print(f"開始遍歷資料夾：{INPUT_DIR}...")
    
    # 第一次遍歷：讀取、轉換並收集所有行
    for root, dirs, files in os.walk(INPUT_DIR):
        folder_name = os.path.basename(root)
        
        if re.match(r'patient\d{4}', folder_name):
            if folder_name not in folder_txt_counts:
                folder_txt_counts[folder_name] = 0

            for filename in files:
                if not filename.endswith('.txt'):
                    continue
                
                folder_txt_counts[folder_name] += 1
                
                file_path = os.path.join(root, filename)
                base_name, _ = os.path.splitext(filename)
                
                try:
                    with open(file_path, 'r') as f:
                        for line in f:
                            line = line.strip()
                            if not line or len(line.split()) < 5:
                                continue
                            
                            try:
                                parts = line.split()
                                x_c = float(parts[1])
                                y_c = float(parts[2])
                                w = float(parts[3])
                                h = float(parts[4])
                                
                                x_min, y_min, x_max, y_max = yolo_to_abs_coords(x_c, y_c, w, h, IMAGE_SIZE)
                                
                                # 格式: patientID_Zindex 0 1 x_min y_min x_max y_max
                                output_line = f"{base_name} 0 1 {x_min} {y_min} {x_max} {y_max}"
                                
                                output_lines.append(output_line)
                                total_annotations += 1
                                
                            except ValueError:
                                continue

                except Exception as e:
                    print(f"讀取檔案 {file_path} 時發生異常: {e}")
                    continue

    # 2. 【新增】對所有收集到的行進行自然排序
    print(f"\n總共處理了 {total_annotations} 條標註。開始排序...")
    if output_lines:
        sorted_lines = sorted(output_lines, key=natural_sort_key)
    else:
        sorted_lines = []

    # 3. 寫入彙整檔案
    try:
        if sorted_lines:
            with open(OUTPUT_FILE, 'w') as out_f:
                for line in sorted_lines:
                    out_f.write(line + '\n')
            print(f"✅ 成功將【排序後】結果寫入彙整檔案：{OUTPUT_FILE}")
        else:
            print("警告：沒有收集到任何有效的標註行，未產生輸出檔案。")
            
    except Exception as e:
        print(f"寫入彙整檔案時發生錯誤: {e}")
        
    # 4. 輸出統計結果
    print("\n--- 每個病人標註影像 (TXT 檔案) 數量統計 ---")
    if folder_txt_counts:
        sorted_patients = sorted(folder_txt_counts.keys())
        for patient_id in sorted_patients:
            count = folder_txt_counts[patient_id]
            print(f"{patient_id}: {count} 張影像有標註")
    else:
        print("未找到符合 'patient####' 格式的子資料夾。")
    print("--------------------------------------------------")


# 執行主函式
aggregate_and_convert_labels()
```

# 過濾一個box txt檔案，每張影像只保留信心指數最高的一個box

```python
# 保留信心指數最高的一個

import os

def keep_highest_confidence_txt(input_path, output_path):
    """
    讀取輸入的 submission 檔案，對每個圖片（以第一欄名稱區分），
    只保留其中信心指數（第三欄）最高的一個標註，並將結果寫入輸出檔案。

    Args:
        input_path (str): 輸入檔案 (e.g., 'submission.txt') 的路徑。
        output_path (str): 輸出檔案 (e.g., 'cleaned_submission.txt') 的路徑。
    """
    # 讀取輸入檔案
    lines = []
    try:
        with open(input_path, 'r') as f:
            for line in f:
                lines.append(line.strip())
        print(f"成功讀取檔案: {input_path}")
    except FileNotFoundError:
        print(f"錯誤：找不到檔案 {input_path}，請檢查路徑是否正確。")
        return
    except Exception as e:
        print(f"讀取檔案時發生錯誤: {e}")
        return

    # --- 核心邏輯：保留最高信心指數的標註 ---
    
    # 使用字典追蹤每個圖片的最佳標註 (最高 confidence)
    best_predictions = {}
    
    for line in lines:
        try:
            parts = line.split()
            patient = parts[0]     # 圖片名稱在第一欄
            # 將信心分數轉換為浮點數
            conf = float(parts[2]) # 信心分數在第三欄
        except (IndexError, ValueError):
            # 忽略格式不正確或欄位不足的行
            continue
            
        # 如果該圖片名稱是第一次出現，或當前的信心分數更高
        if patient not in best_predictions or conf > best_predictions[patient]['conf']:
            # 更新為當前最高的信心分數和對應的整行內容
            best_predictions[patient] = {'conf': conf, 'line': line}

    # 提取所有被選定的最佳標註行
    kept_lines = [item['line'] for item in best_predictions.values()]
    
    # 計算統計
    total_lines = len(lines)
    deleted = total_lines - len(kept_lines)

    # 輸出統計結果
    print("-" * 30)
    print(f"總行數 (原始標註數量): {total_lines}")
    print(f"刪除了重複且信心較低的 bounding boxes: {deleted} 個")
    print(f"保留了唯一的最佳標註 (每個圖片一個): {len(kept_lines)} 個")
    print("-" * 30)

    # 寫入清理後的檔案
    try:
        with open(output_path, 'w') as f:
            for line in kept_lines:
                f.write(line + '\n')
        print(f"已輸出清理後的檔案: {output_path}")
    except Exception as e:
        print(f"寫入檔案時發生錯誤: {e}")


```

# 去頭去尾函數
- 2025-11-16

        函式說明與使用建議

        核心邏輯：每個病人只保留「最大連續群組」（z_index 間隔 ≤ gap_threshold 的最長段落），這假設主體瓣膜是連續的，孤立 box（前/後間隔 >3 空白切片）是假陽性。這樣可減少 FP，提升 Precision，而不大幅損 Recall。
        參數調整：
        gap_threshold=3：CT 切片間隔 1-3 正常（心臟運動），>3 視為不連續。依視覺化調整（e.g., 5 若間隔更大）。
        min_group_size=2：單一 box 刪除，2+ 保留。設 1 則保留所有但過濾孤立。

        輸出：新 txt 完全符合原格式，按 conf 降序排序（競賽評分會依此逐一配對 GT）。無需後處理。
        檢查：用原 plot_z_axis_distribution_ZY/ZX 函式，輸入 filtered_df，確認不連續 box 已刪除。

```python
import pandas as pd
import numpy as np

def filter_continuous_boxes(df_clean, output_file, gap_threshold=3, min_group_size=2):
    """
    過濾不連續的 bounding box，保留每個病人最大連續群組。
    
    邏輯：
    - 按 patient_id 分組。
    - 對每個病人：排序 z_index，偵測連續群組（相鄰 z_index 差 <= gap_threshold）。
    - 保留每個群組長度 >= min_group_size 的群組，並選擇最大群組（主體）。
    - 合併所有保留 row，按 confidence 降序排序。
    - 輸出新 txt 檔案，格式與原 submission 相同。
    
    Args:
        df_clean (pd.DataFrame): 經過 preprocess_data 處理的 DataFrame。
        output_file (str): 輸出新 txt 檔案路徑。
        gap_threshold (int): 連續閾值，間隔 > 此值視為新群組 (預設 3)。
        min_group_size (int): 最小群組大小，< 此值刪除 (預設 2)。
    
    Returns:
        pd.DataFrame: 過濾後的 DataFrame。
    """
    if df_clean is None or df_clean.empty:
        print("輸入 DataFrame 為空，無法過濾。")
        return None
    
    filtered_rows = []
    
    # 按 patient_id 分組
    for patient_id, patient_df in df_clean.groupby('patient_id'):
        if patient_df.empty:
            continue
        
        # 排序 z_index
        patient_df = patient_df.sort_values('z_index').reset_index(drop=True)
        z_indices = patient_df['z_index'].values
        unique_z = np.unique(z_indices)  # 去重，若同一 z 有重複 box (罕見)
        if len(unique_z) < min_group_size:
            continue  # 太少 box，全部刪除
        
        # 偵測連續群組
        groups = []
        current_group = [unique_z[0]]
        for i in range(1, len(unique_z)):
            if unique_z[i] - unique_z[i-1] <= gap_threshold:
                current_group.append(unique_z[i])
            else:
                if len(current_group) >= min_group_size:
                    groups.append(current_group)
                current_group = [unique_z[i]]
        if len(current_group) >= min_group_size:
            groups.append(current_group)
        
        if not groups:
            continue  # 無有效群組
        
        # 選擇最大群組 (主體)
        max_group = max(groups, key=len)
        z_in_group = set(max_group)
        
        # 保留該群組的 row (若同一 z 多 box，保留所有)
        patient_filtered = patient_df[patient_df['z_index'].isin(z_in_group)]
        filtered_rows.append(patient_filtered)
    
    if not filtered_rows:
        print("無保留的 box，輸出空檔案。")
        pd.DataFrame().to_csv(output_file, sep=' ', index=False, header=False)
        return pd.DataFrame()
    
    # 合併所有保留 row
    df_filtered = pd.concat(filtered_rows, ignore_index=True)
    
    # 按 confidence 降序排序 (全域排序，符合競賽評分流程)
    df_filtered = df_filtered.sort_values('confidence', ascending=False).reset_index(drop=True)
    
    # 寫入新 txt 檔案：filename class confidence x1 y1 x2 y2
    df_output = df_filtered[['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']].copy()
    df_output.to_csv(output_file, sep=' ', index=False, header=False, float_format='%.4f')
    
    print(f"過濾完成！保留 {len(df_filtered)} 個 box，原 {len(df_clean)} 個。")
    print(f"新檔案已儲存至: {output_file}")
    
    return df_filtered

# --- 使用範例 ---
# 假設你已載入 df_clean 和 DWBF_txt
# filtered_df = filter_continuous_boxes(df_clean, 'filtered_predictions.txt', gap_threshold=3, min_group_size=2)

# # 若要視覺化檢查 (使用原 plot 函式)
# target_patient = 'patient0100'  # 替換成你要檢查的病人
# plot_z_axis_distribution_ZY(filtered_df, target_patient)
# plot_z_axis_distribution_ZX(filtered_df, target_patient)
```

```python
import pandas as pd
import numpy as np

def replace_outlier_boxes(df_clean, output_file, window_size=3, std_threshold=2.0, axis='both'):
    """
    替換 outlier bounding box 位置，使用前後 window_size 個 box 的平均位置。
    
    邏輯：
    - 按 patient_id 分組。
    - 對每個病人：排序 z_index，計算每個 box 的 y_center / x_center 移動平均 (window_size)。
    - 偵測 outlier：位置偏差 > std_threshold * 移動 std。
    - 替換：用前後 window_size 個的平均 x_center / y_center 替換 (維持框大小與 conf)。
    - 合併所有 row，按 confidence 降序排序。
    - 輸出新 txt 檔案，格式與原 submission 相同。
    
    Args:
        df_clean (pd.DataFrame): 經過 preprocess_data 處理的 DataFrame。
        output_file (str): 輸出新 txt 檔案路徑。
        window_size (int): 前後平均視窗大小 (預設 3: 前1 + 當前 + 後1)。
        std_threshold (float): Outlier 偵測閾值 (預設 2.0: >2 std 視為 outlier)。
        axis (str): 'y' (Y軸), 'x' (X軸), 'both' (兩者都檢查, 預設)。
    
    Returns:
        pd.DataFrame: 替換後的 DataFrame。
    """
    if df_clean is None or df_clean.empty:
        print("輸入 DataFrame 為空，無法替換。")
        return None
    
    df_modified = df_clean.copy()
    
    # 按 patient_id 分組
    for patient_id, patient_df in df_modified.groupby('patient_id'):
        if len(patient_df) < window_size:
            continue  # 太少 box，跳過
        
        # 排序 z_index
        patient_df = patient_df.sort_values('z_index').reset_index(drop=True)
        
        # 計算移動平均與 std (用於偵測 outlier)
        centers = {'y': patient_df['y_center'].values, 'x': patient_df['x_center'].values}
        rolling_mean = {'y': pd.Series(centers['y']).rolling(window=window_size, center=True).mean().fillna(method='bfill').fillna(method='ffill').values,
                        'x': pd.Series(centers['x']).rolling(window=window_size, center=True).mean().fillna(method='bfill').fillna(method='ffill').values}
        rolling_std = {'y': pd.Series(centers['y']).rolling(window=window_size, center=True).std().fillna(0).values,
                       'x': pd.Series(centers['x']).rolling(window=window_size, center=True).std().fillna(0).values}
        
        # 偵測並替換 outlier
        for idx in patient_df.index:
            replace_y = False
            replace_x = False
            
            if axis in ['y', 'both']:
                y_dev = abs(centers['y'][idx] - rolling_mean['y'][idx])
                if y_dev > std_threshold * rolling_std['y'][idx]:
                    replace_y = True
            
            if axis in ['x', 'both']:
                x_dev = abs(centers['x'][idx] - rolling_mean['x'][idx])
                if x_dev > std_threshold * rolling_std['x'][idx]:
                    replace_x = True
            
            if replace_y or replace_x:
                # 計算前後 window_size 平均 (排除當前)
                start_idx = max(0, idx - (window_size // 2))
                end_idx = min(len(patient_df), idx + (window_size // 2) + 1)
                neighbors_y = centers['y'][start_idx:end_idx]
                neighbors_y = neighbors_y[neighbors_y != centers['y'][idx]]  # 排除當前
                neighbors_x = centers['x'][start_idx:end_idx]
                neighbors_x = neighbors_x[neighbors_x != centers['x'][idx]]
                
                if len(neighbors_y) > 0:
                    new_y_center = np.mean(neighbors_y)
                    # 維持框高度：y1, y2 = center ± half_height
                    half_h = (patient_df.at[idx, 'y_max'] - patient_df.at[idx, 'y_min']) / 2
                    patient_df.at[idx, 'y1'] = new_y_center - half_h
                    patient_df.at[idx, 'y2'] = new_y_center + half_h
                    replace_y = True
                
                if len(neighbors_x) > 0:
                    new_x_center = np.mean(neighbors_x)
                    half_w = (patient_df.at[idx, 'x_max'] - patient_df.at[idx, 'x_min']) / 2
                    patient_df.at[idx, 'x1'] = new_x_center - half_w
                    patient_df.at[idx, 'x2'] = new_x_center + half_w
                    replace_x = True
                
                if replace_y or replace_x:
                    print(f"病人 {patient_id}, z_index {patient_df.at[idx, 'z_index']}: 替換 outlier (Y: {replace_y}, X: {replace_x})")
        
        # 更新 df_modified
        df_modified.loc[df_modified.index.isin(patient_df.index), :] = patient_df
    
    # 按 confidence 降序排序 (全域)
    df_filtered = df_modified.sort_values('confidence', ascending=False).reset_index(drop=True)
    
    # 寫入新 txt 檔案：filename class confidence x1 y1 x2 y2
    df_output = df_filtered[['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']].copy()
    
    # xy to int
    df_output['x1'] = df_output['x1'].astype(int)
    df_output['y1'] = df_output['y1'].astype(int)
    df_output['x2'] = df_output['x2'].astype(int)
    df_output['y2'] = df_output['y2'].astype(int)

    
    df_output.to_csv(output_file, sep=' ', index=False, header=False, float_format='%.4f')
    
    print(f"替換完成！處理 {len(df_modified)} 個 box。")
    print(f"新檔案已儲存至: {output_file}")
    
    return df_filtered

# --- 使用範例 ---
# 假設你已載入 df_clean 和 DWBF_txt
# filtered_df = replace_outlier_boxes(df_clean, 'replaced_predictions.txt', window_size=3, std_threshold=2.0, axis='both')

# # 若要視覺化檢查 (使用原 plot 函式)
# target_patient = 'patient0054'  # 從圖中替換
# plot_z_axis_distribution_ZY(filtered_df, target_patient)
# plot_z_axis_distribution_ZX(filtered_df, target_patient)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import numpy as np
import os
from copy import deepcopy

# 右鍵選單問題
import tkinter as tk
from tkinter import Menu as TkMenu


# 存檔問題
from matplotlib.widgets import Button

# -- 中文字型問題 --
#import matplotlib
#matplotlib.rc('font', family='Microsoft JhengHei')

# --- 1. 數據讀取與解析函式 --- 重複使用

def load_yolo_submission(file_path):
    """
    讀取 YOLO 競賽提交檔，並解析成 DataFrame。
    欄位順序：[檔名, 類別, 信心指數, x1, y1, x2, y2]
    """
    try:
        # 由於數據沒有列名，我們手動指定它們
        df = pd.read_csv(
            file_path,
            sep=' ',
            header=None,
            names=['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']
        )
        return df
    except FileNotFoundError:
        print(f"錯誤：找不到檔案 {file_path}")
        return None
    except Exception as e:
        print(f"讀取檔案時發生錯誤: {e}")
        return None
        
# --- 2. 數據處理與轉換函式 --- 重複使用

def preprocess_data(df):
    """
    從 DataFrame 中提取病人ID和影像Z軸編號。
    """
    if df is None:
        return None

    # 提取病人ID (patient00##) 和 Z軸編號 (####)
    # 使用正規表達式提取：
    # patient(00\d{2}) -> 擷取 00xx 部分
    # _(\d{4}) -> 擷取四位數字部分
    df['patient_id'] = df['filename'].str.extract(r'(patient\d{4})')
    df['z_index'] = df['filename'].str.extract(r'_(\d{4})').astype(int)

    # 計算預測框的 Y 軸高度（用於視覺化）
    # y_center: 預測框的中心點 Y 座標
    # y_min: 預測框的最小 Y 座標 (y1)
    # y_max: 預測框的最大 Y 座標 (y2)
    df['y_min'] = df[['y1', 'y2']].min(axis=1) # 確保 y_min 總是較小的值
    df['y_max'] = df[['y1', 'y2']].max(axis=1) # 確保 y_max 總是較大的值
    df['y_center'] = (df['y_min'] + df['y_max']) / 2
    # X_
    df['x_min'] = df[['x1', 'x2']].min(axis=1)
    df['x_max'] = df[['x1', 'x2']].max(axis=1)
    df['x_center'] = (df['x_min'] + df['x_max']) / 2
    
    return df

        
        
        
class BoxEditorGUI:
    def __init__(self, df_clean, output_dir="edited_submissions"):
        self.df = df_clean.copy()
        self.original_df = df_clean.copy()
        self.output_dir = output_dir
        os.makedirs(output_dir, exist_ok=True)
        

        
        
        # 提取所有病人
        self.patients = sorted(self.df['patient_id'].unique())
        self.current_patient_idx = 0
        
        # 編輯歷史 (支援 undo/redo)
        self.history = []
        self.history_idx = -1
        self.max_history = 50
        
        # 複製緩衝區
        self.clipboard = None
        
        # 圖表
        self.fig, (self.ax_y, self.ax_x) = plt.subplots(1, 2, figsize=(20, 10))
        self.fig.canvas.manager.set_window_title("Box check tool")
        self.fig.suptitle("Click red center to 'Del' | left click copy/past | Ctrl+Z undo", fontsize=12)
        
        # 事件綁定
        self.cid_click = self.fig.canvas.mpl_connect('button_press_event', self.on_click)
        #self.cid_key = self.fig.canvas.mpl_connect('key_press_event', self.on_key)
        # 修改 on_key
        self.cid_key = self.fig.canvas.mpl_connect('key_press_event', self.on_key)
        
        # 右鍵選單
        self.context_menu = None
        
        self.redraw()
        plt.show()

    def get_current_patient(self):
        return self.patients[self.current_patient_idx]
    
    def get_current_data(self):
        pid = self.get_current_patient()
        return self.df[self.df['patient_id'] == pid].copy().sort_values('z_index')
    
    def save_history(self):
        """儲存當前狀態到歷史"""
        current_state = self.df.copy()
        self.history = self.history[:self.history_idx + 1]
        self.history.append(current_state)
        if len(self.history) > self.max_history:
            self.history.pop(0)
        self.history_idx = len(self.history) - 1

    def undo(self):
        if self.history_idx > 0:
            self.history_idx -= 1
            self.df = self.history[self.history_idx].copy()
            self.redraw()
    
    def redo(self):
        if self.history_idx < len(self.history) - 1:
            self.history_idx += 1
            self.df = self.history[self.history_idx].copy()
            self.redraw()

    def redraw(self):
        self.ax_y.clear()
        self.ax_x.clear()
        
        data = self.get_current_data()
        if data.empty:
            self.ax_y.text(0.5, 0.5, "無資料", transform=self.ax_y.transAxes, ha='center')
            self.ax_x.text(0.5, 0.5, "無資料", transform=self.ax_x.transAxes, ha='center')
            self.fig.canvas.draw()
            return
        
        z = data['z_index'].values
        y_min, y_max = data['y_min'].values, data['y_max'].values
        y_center = data['y_center'].values
        x_min, x_max = data['x_min'].values, data['x_max'].values
        x_center = data['x_center'].values
        conf = data['confidence'].values
        
        # === Y/Z 圖 ===
        for zi, ymin, ymax, c in zip(z, y_min, y_max, conf):
            self.ax_y.plot([zi, zi], [ymin, ymax], color='blue', alpha=c, lw=1.5)
            self.ax_y.scatter(zi, (ymin+ymax)/2, s=50*c + 10, c='red', zorder=5, picker=True)
        
        self.ax_y.set_xlabel("Z-Index (CT Slice)")
        self.ax_y.set_ylabel("Y Pixel (Top → Bottom)")
        self.ax_y.set_title(f"Y/Z Distribution - {self.get_current_patient()}")
        self.ax_y.invert_yaxis()
        self.ax_y.grid(True, alpha=0.3)
        
        # --- 【修正處】強制設定 Y 軸範圍並反轉 ---
        # 即使共享 Y 軸，也強制設定範圍並反轉，確保所有子圖統一
        y_lim_min = 150   #統一設定Y軸上下限
        y_lim_max = 450   #統一設定Y軸上下限
        self.ax_y.set_ylim(y_lim_max, y_lim_min) # 注意：這裡的 max, min 順序是反的！
                                         # (y_lim_max 在第一個參數，讓它變成 Y 軸頂部)
        
        
        
        
        # === X/Z 圖 ===
        for zi, xmin, xmax, c in zip(z, x_min, x_max, conf):
            self.ax_x.plot([zi, zi], [xmin, xmax], color='green', alpha=c, lw=1.5)
            self.ax_x.scatter(zi, (xmin+xmax)/2, s=50*c + 10, c='red', zorder=5, picker=True)
        
        self.ax_x.set_xlabel("Z-Index (CT Slice)")
        self.ax_x.set_ylabel("X Pixel (Left → Right)")
        self.ax_x.set_title(f"X/Z Distribution - {self.get_current_patient()}")
        self.ax_x.grid(True, alpha=0.3)
        
                # --- 【修正處】強制設定 Y 軸範圍並反轉 ---
        # 即使共享 Y 軸，也強制設定範圍並反轉，確保所有子圖統一
        x_lim_min = 150   #統一設定Y軸上下限
        x_lim_max = 450   #統一設定Y軸上下限
        self.ax_x.set_ylim(x_lim_max, x_lim_min) # 注意：這裡的 max, min 順序是反的！
                                         # (y_lim_max 在第一個參數，讓它變成 Y 軸頂部)
                                         
                             
        # 存檔案扭
        # 另外增加一個save按鍵
        #from matplotlib.widgets import Button
        ax_save = plt.axes([0.43, 0.01, 0.18, 0.06])
        self.btn_save = Button(ax_save, 'Save current patient\n(Ctrl+S)', color='#90EE90')
        self.btn_save.on_clicked(lambda x: self.save_current_patient())
        
        # 狀態列
        self.fig.text(0.01, 0.01, f"Patient: {self.get_current_patient()} | Nbox: {len(data)} | "
                                 f"history: {self.history_idx+1}/{len(self.history)}",
                     fontsize=10, transform=self.fig.transFigure)
        
        self.fig.canvas.draw()

#    def on_key(self, event):
#        if event.key == 'left' and self.current_patient_idx > 0:
#            self.current_patient_idx -= 1
#            self.redraw()
#        elif event.key == 'right' and self.current_patient_idx < len(self.patients)-1:
#            self.current_patient_idx += 1
#            self.redraw()
#        elif event.key in ['ctrl+z', 'cmd+z']:
#            self.undo()
#        elif event.key in ['ctrl+y', 'cmd+y']:
#            self.redo()
#        elif event.key == 's':
#            self.save_current_patient()

    def on_key(self, event):
        if event.key in ['ctrl+s', 'cmd+s']:
            self.save_current_patient()
            try:
                if hasattr(event, 'guiEvent') and event.guiEvent:
                    event.guiEvent.Skip()
            except:
                pass
            return

        if event.key == 'left' and self.current_patient_idx > 0:
            self.current_patient_idx -= 1
            self.redraw()
        elif event.key == 'right' and self.current_patient_idx < len(self.patients)-1:
            self.current_patient_idx += 1
            self.redraw()
        elif event.key in ['ctrl+z', 'cmd+z']:
            self.undo()
        elif event.key in ['ctrl+y', 'cmd+y']:
            self.redo()
            

    def on_click(self, event):
        if event.inaxes not in [self.ax_y, self.ax_x]:
            return
        if event.button == 1:  # 左鍵：刪除
            self.delete_nearest_box(event)
        elif event.button == 3:  # 右鍵：選單
            self.show_context_menu(event)

    def delete_nearest_box(self, event):
        data = self.get_current_data()
        z = data['z_index'].values
        if event.inaxes == self.ax_y:
            centers = data['y_center'].values
        else:
            centers = data['x_center'].values
        
        # 找最近的點
        dist = np.abs(z - event.xdata) + np.abs(centers - event.ydata) * 0.1
        idx = np.argmin(dist)
        global_idx = data.index[idx]
        
        self.save_history()
        self.df = self.df.drop(global_idx)
        print(f"刪除: {data.iloc[idx]['filename']} (conf={data.iloc[idx]['confidence']:.4f})")
        self.redraw()

#    def show_context_menu(self, event):
#        from matplotlib.widgets import Menu
#        data = self.get_current_data()
#        z = data['z_index'].values
#        centers = data['y_center' if event.inaxes == self.ax_y else 'x_center'].values
#        dist = np.abs(z - event.xdata) + np.abs(centers - event.ydata) * 0.1
#        idx = np.argmin(dist)
#        self.selected_row = data.iloc[idx]
#        self.selected_global_idx = data.index[idx]
#        
#        menu = Menu(self.fig, [
#            ("copy this box", self.copy_box),
#            ("past on here (replace)", lambda: self.paste_box(replace=True)),
#            ("past as new box", lambda: self.paste_box(replace=False)),
#        ])
#        menu.popup(event)

    #原因是：matplotlib.widgets.Menu 從 3.4 版開始就已經被移除（deprecated → removed），現在官方已不再提供這個類別。
#    def show_context_menu(self, event):
#        # 找出最近的 box
#        data = self.get_current_data()
#        if data.empty:
#            return
#        z = data['z_index'].values
#        centers = data['y_center' if event.inaxes == self.ax_y else 'x_center'].values
#        dist = np.abs(z - event.xdata) + np.abs(centers - event.ydata) * 0.1
#        idx = np.argmin(dist)
#        self.selected_row = data.iloc[idx].copy()
#        self.selected_global_idx = data.index[idx]
#
#        # 使用 tkinter 右鍵選單（跨平台、穩定）
#        try:
#            menu = plt.tk.Tk()
#            menu.withdraw()  # 隱藏主窗口
#            rmenu = plt.tk.Menu(menu, tearoff=0)
#            rmenu.add_command(label="複製此 box", command=self.copy_box)
#            rmenu.add_command(label="貼上並替換此位置", command=lambda: self.paste_box(replace=True))
#            rmenu.add_command(label="貼上為新 box", command=lambda: self.paste_box(replace=False))
#            rmenu.add_separator()
#            rmenu.add_command(label="刪除此 box", command=lambda: self.delete_selected_box())
#            
#            # 捕捉關閉事件，避免殘留
#            try:
#                rmenu.tk_popup(int(event.x_root), int(event.y_root))
#            finally:
#                rmenu.grab_release()
#        except Exception as e:
#            print("右鍵選單錯誤（tkinter）:", e)
#            # 後備：用 print 提示
#            print("請用左鍵刪除，右鍵功能暫不可用")
 


    def show_context_menu(self, event):
        """右鍵選單：使用純 tkinter（不依賴 matplotlib.tk）"""
        data = self.get_current_data()
        if data.empty:
            return

        # 找最近的 box
        z = data['z_index'].values
        centers = data['y_center' if event.inaxes == self.ax_y else 'x_center'].values
        dist = np.abs(z - event.xdata) + np.abs(centers - event.ydata) * 0.1
        idx = np.argmin(dist)
        self.selected_row = data.iloc[idx].copy()
        self.selected_global_idx = data.index[idx]

        # === 純 tkinter 右鍵選單（最穩定）===
        root = tk.Tk()
        root.withdraw()  # 隱藏主窗口
        menu = TkMenu(root, tearoff=0)
        menu.add_command(label="複製此 box", command=self.copy_box)
        menu.add_command(label="貼上並替換此位置", command=lambda: self.paste_box(replace=True))
        menu.add_command(label="貼上為新 box", command=lambda: self.paste_box(replace=False))
        menu.add_separator()
        menu.add_command(label="刪除此 box", command=self.delete_selected_box)

        try:
            # 這兩行是關鍵：正確轉換座標
            menu.tk_popup(int(event.x_root), int(event.y_root))
        finally:
            menu.grab_release()
            root.destroy()  # 一定要銷毀，避免殘留
            
            
 
 
    def delete_selected_box(self):
        if hasattr(self, 'selected_global_idx'):
            self.save_history()
            self.df = self.df.drop(self.selected_global_idx)
            print(f"已刪除: {self.selected_row['filename']}")
            self.redraw()

    def copy_box(self):
        self.clipboard = self.selected_row.copy()
        print(f"已複製: {self.clipboard['filename']} (conf={self.clipboard['confidence']:.4f})")

    def paste_box(self, replace=True):
        if self.clipboard is None:
            print("剪貼簿為空！")
            return
        
        data = self.get_current_data()
        z_click = int(round(self.selected_row['z_index']))  # 點擊的 z
        new_row = self.clipboard.copy()
        new_row['z_index'] = z_click
        new_row['filename'] = new_row['filename'].str.replace(r'_\d{4}', f'_{z_click:04d}', regex=True)
        
        self.save_history()
        if replace:
            # 替換該 z 的所有 box
            self.df = self.df[~((self.df['patient_id'] == self.get_current_patient()) & 
                               (self.df['z_index'] == z_click))]
        
        self.df = pd.concat([self.df, pd.DataFrame([new_row])], ignore_index=True)
        print(f"past ok on z={z_click}")
        self.redraw()

    def save_current_patient(self):
        pid = self.get_current_patient()
        patient_data = self.df[self.df['patient_id'] == pid].copy()
        if patient_data.empty:
            print(f"病人 {pid} 無資料，跳過儲存。")
            return
        
        # 合併所有病人（維持原排序）
        other_patients = self.df[self.df['patient_id'] != pid]
        final_df = pd.concat([other_patients, patient_data], ignore_index=True)
        final_df = final_df.sort_values('confidence', ascending=False)
        
        # 輸出欄位
        output_cols = ['filename', 'class', 'confidence', 'x1', 'y1', 'x2', 'y2']
        output_df = final_df[output_cols].copy()
        output_df['x1'] = output_df['x1'].astype(int)
        output_df['y1'] = output_df['y1'].astype(int)
        output_df['x2'] = output_df['x2'].astype(int)
        output_df['y2'] = output_df['y2'].astype(int)
        
        output_path = os.path.join(self.output_dir, f"edited_submission.txt")
        output_df.to_csv(output_path, sep=' ', index=False, header=False, float_format='%.4f')
        print(f"已儲存: {output_path} (含 {len(patient_data)} 個 box)")

# ======================
# 使用方式
# ======================
# 1. 載入你的 fused txt
#df_raw = load_yolo_submission("fused_predictions_v9e-W7-1112_V2_order_refine-01.txt")
df_raw = load_yolo_submission("predict_txt/detr-V4.txt")
df_clean = preprocess_data(df_raw)

# 2. 啟動 GUI
editor = BoxEditorGUI(df_clean)

# 3. 操作：
#    ← → 切換病人
#    左鍵點擊中心點 → 刪除
#    右鍵 → 複製/貼上
#    Ctrl+S → 儲存目前病人
#    Ctrl+Z / Y → 復原/重做
```
